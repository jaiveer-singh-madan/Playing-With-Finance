```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from eodhd import APIClient
from dotenv import load_dotenv
import os
import time
```

```python
load_dotenv()
token = os.getenv("EOD_API_KEY")
api = APIClient(token)
```

```python
tickers=["ADYEN.AS"]
start_date='2026-01-01'

for ticker in tickers:
    data=api.get_eod_historical_stock_market_data(ticker, from_date=start_date)
    df = pd.DataFrame(data)
    df['date'] = pd.to_datetime(df['date'])
    df.set_index('date', inplace=True)
    df.sort_index(inplace=True)
    returns = df['close'].pct_change()
    daily_volatility = returns.std() * 100
    annualised_volatility = returns.std() * (252 ** 0.5) * 100
    print(annualised_volatility)
    '''
    plt.figure(figsize=(10, 5))
    plt.plot(df.index, df['close'], label=ticker)
    plt.title(f'{ticker} Closing Prices')
    plt.xlabel('Date')
    plt.ylabel('Closing Price')
    plt.legend()
    plt.grid()
    plt.show()
    '''
    time.sleep(1)  # To avoid hitting API rate limits
```
